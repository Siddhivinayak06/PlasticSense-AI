# 🚀 PlasticSense AI - Project Initialization

Welcome to **PlasticSense AI**, an IEEE Final Year Project focusing on Computer Vision and Object Detection using YOLOv11.

This notebook is designed to initialize the project environment in Google Colab. It performs the following tasks:
1. Mounts Google Drive.
2. Creates the `PlasticSense_AI` project directory and required folder structure.
3. Installs required libraries (YOLOv11, PyTorch, OpenCV, etc.).
4. Sets up reusable configuration variables using `pathlib`.
5. Initializes a logging system.
6. Displays project information and verifies the folder structure.
7. Saves configurations into a `config.py` file for modular use across other scripts.

```python
# 1. Mount Google Drive
from google.colab import drive
import os

# Mount Google Drive to save project files persistently
drive.mount('/content/drive')
```

```python
# 2, 5. Project Folder Setup and Pathlib Configurations
from pathlib import Path

# Define the base project directory in Google Drive
PROJECT_NAME = "PlasticSense_AI"
BASE_DIR = Path('/content/drive/MyDrive') / PROJECT_NAME

# Define the folder structure
DIRS = {
    "notebooks": BASE_DIR / "notebooks",
    "datasets": BASE_DIR / "datasets",
    "configs": BASE_DIR / "configs",
    "models": BASE_DIR / "models",
    "exports": BASE_DIR / "exports",
    "logs": BASE_DIR / "logs",
    "results": BASE_DIR / "results",
    "results_training": BASE_DIR / "results" / "training",
    "results_evaluation": BASE_DIR / "results" / "evaluation",
    "results_inference": BASE_DIR / "results" / "inference",
}

# Create the directories
def create_project_structure(dirs):
    print("Initializing Project Structure...")
    for name, path in dirs.items():
        path.mkdir(parents=True, exist_ok=True)
        print(f"Created/Verified: {path}")
    print("Project structure initialization complete.\n")

create_project_structure(DIRS)
```

```python
# 6. Logging System
import logging
import sys

def setup_logger(log_file_path):
    """Sets up a robust logger that logs to both the Colab console and a persistent file."""
    logger = logging.getLogger("PlasticSense_AI")
    logger.setLevel(logging.INFO)
    
    # Clear existing handlers to prevent duplicate logs if cell is run multiple times
    if logger.hasHandlers():
        logger.handlers.clear()
        
    # Create handlers
    c_handler = logging.StreamHandler(sys.stdout)
    f_handler = logging.FileHandler(log_file_path)
    
    # Create formatters and add it to handlers
    log_format = logging.Formatter('%(asctime)s | %(name)s | %(levelname)s | %(message)s', datefmt='%Y-%m-%d %H:%M:%S')
    c_handler.setFormatter(log_format)
    f_handler.setFormatter(log_format)
    
    # Add handlers to the logger
    logger.addHandler(c_handler)
    logger.addHandler(f_handler)
        
    return logger

# Initialize logger
log_file = DIRS["logs"] / "project_setup.log"
logger = setup_logger(log_file)
logger.info("Logging system initialized successfully. Logs are being saved to Google Drive.")
```

```python
# 7, 8. Display Project Information and Verify Structure
def verify_and_display_info(dirs, logger, base_dir):
    logger.info("--- PlasticSense AI Project Summary ---")
    logger.info(f"Base Directory: {base_dir}")
    
    missing_dirs = []
    logger.info("Verifying Folder Structure:")
    for name, path in dirs.items():
        if path.exists():
            logger.info(f"  [✓] {name:<20} -> {path}")
        else:
            logger.error(f"  [✗] {name:<20} -> {path}")
            missing_dirs.append(name)
            
    if not missing_dirs:
        logger.info("All project directories verified successfully.")
    else:
        logger.warning(f"Missing directories: {missing_dirs}")

verify_and_display_info(DIRS, logger, BASE_DIR)
```

```python
# 9. Save all configuration into config.py style variables
import textwrap

def save_config_to_py(dirs, base_dir, config_dir, logger):
    config_path = config_dir / "config.py"
    
    # Generate the config.py content
    config_content = textwrap.dedent(f'''\
    """
    PlasticSense AI - Core Configuration File
    Auto-generated by Project Setup Notebook.
    """
    from pathlib import Path

    # Base Project Directory (Google Drive path)
    BASE_DIR = Path("{base_dir}")

    # Core Project Directories
    DIRS = {{
        "notebooks": BASE_DIR / "notebooks",
        "datasets": BASE_DIR / "datasets",
        "configs": BASE_DIR / "configs",
        "models": BASE_DIR / "models",
        "exports": BASE_DIR / "exports",
        "logs": BASE_DIR / "logs",
        "results": BASE_DIR / "results",
        "results_training": BASE_DIR / "results" / "training",
        "results_evaluation": BASE_DIR / "results" / "evaluation",
        "results_inference": BASE_DIR / "results" / "inference",
    }}
    
    # Model Training Configuration Defaults (YOLOv11)
    TRAIN_CONFIG = {{
        "model": "yolo11n.pt", # Default YOLOv11 nano model
        "epochs": 100,
        "batch_size": 16,
        "imgsz": 640,
        "device": "cuda:0", # Use GPU
        "project": str(DIRS["results_training"]),
        "name": "plastic_detection_exp1",
    }}
    ''')
    
    with open(config_path, "w") as f:
        f.write(config_content)
        
    logger.info(f"Configuration successfully saved to {config_path}")

save_config_to_py(DIRS, BASE_DIR, DIRS["configs"], logger)

logger.info("✅ Project Setup Complete! You are ready to start building PlasticSense AI.")
```
